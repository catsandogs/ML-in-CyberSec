#Пункт 2. Выбрать и обосновать метрику для измерения качества.

Прежде чем мы сможем определить верные метрики, следует выполнить разведочный анализ (EDA) - поэтому этот пункт я пишу после того, как закончил с EDA. Таким образом, было выявлено, что классы в датасете сильно несбалансированы: BENIGN составляет 52%, а доля редких атак (например, DNS_Tunneling) - около 2%. В таких условиях обычная accuracy неприменима, так как модель, всегда предсказывающая BENIGN, получит 52% точности, но не обнаружит ни одной атаки. Хотя accuracy была бы хороша, если бы мы решали бинарную классификацию.

Поэтому основной метрикой выбирается **Macro F1-score.** При дисбалансе классов способ усреднения критически важен: Macro F1 считает метрику для каждого из 13 классов отдельно и усредняет их с одинаковым весом. В отличие от Micro F1, где частые классы «задавливают» редкие, Macro F1 заставляет модель искать признаки для каждого типа атак, включая миноритарные.

Дополнительно используются **Per-class Precision и Recall** - чтобы видеть, какие именно атаки модель пропускает. Также рассчитывается **MCC (Matthews Correlation Coefficient)** - наиболее сбалансированная метрика, учитывающая все четыре компонента матрицы ошибок и не завышающая результат за счёт преобладающего класса.

В качестве стресс-метрики проверяется F1 на подвыборке классов с долей менее 3%. Если он падает ниже 0.5 - модель «забыла» про редкие угрозы. Пороги: норма (Macro F1 > 0.85, MCC > 0.80), предупреждение (Macro F1 < 0.80, MCC < 0.75), критический (Macro F1 < 0.70 - модель непригодна для ИБ).

А flag_anomaly_score ищет странные комбинации флагов (например, сбросы RST или SYN-флуд без завершения). Поэтому логично: если много ACK, то трафик выглядит здоровым, и аномальность падает.

# Feature engineering (Или не совсем?)
Далее мне стало интересно проанализировать как коррелируют те или иные признаки с самим фактом атаки. Для этого нужно создать новый признак "is_attack", где 0 - это отсутсвие атаки(BENIGN), а 1 - наборот наличие одной из атак, которые могут быть в датасете. Понимаю, что мы решаем задачу многоклассовой классификации и делать бинарные признаки тут может быть неуместно, но данный анализ может помочь для построение модели в будущем и покажет на какие признаки обратить внимание.

```python
#  создаём временный бинарный признак
df_eda = df.copy()
df_eda['is_attack'] = (df_eda['label'] != 'BENIGN').astype(int)

# выбираем числовые признаки, исключаем идентификаторы, время и нашу новую метку
features_to_corr = [c for c in df_eda.select_dtypes(include=np.number).columns
                    if c not in ['src_port', 'dst_port', 'timestamp', 'is_attack']]

# считаем корреляцию Спирмена
corr_with_target = df_eda[features_to_corr].corrwith(df_eda['is_attack'], method='spearman').sort_values(ascending=False)

plt.figure(figsize=(12, 10))
top_corr = pd.concat([corr_with_target.head(10), corr_with_target.tail(10)])
sns.barplot(x=top_corr.values, y=top_corr.index, palette='coolwarm')
plt.title('топ признаков по корреляции с фактом атаки (Спирмен)')
plt.xlabel('коэффициент корреляции')
plt.ylabel('признак')
plt.axvline(0, color='black', linestyle='--')
plt.tight_layout()
plt.show()


print("Топ признаков, которые сильнее всего связаны с атакой:")
print(corr_with_target.head(10))
```

Далее я понял, что следует смотреть на те признаки, в которых наибольший коэффициент корреляции по модулю.

```python
corr_with_target = df_eda[features_to_corr].corrwith(df_eda['is_attack'], method='spearman')

# сортируем по модулю корреляции
top_20_features = corr_with_target.abs().sort_values(ascending=False).head(20).index

top_20_values = corr_with_target[top_20_features].sort_values()

plt.figure(figsize=(12, 10))
sns.barplot(x=top_20_values.values, y=top_20_values.index, palette='coolwarm')
plt.title('Топ-20 признаков по СИЛЕ связи с фактом атаки (Спирмен)')
plt.xlabel('Коэффициент корреляции')
plt.ylabel('Признак')
plt.axvline(0, color='black', linestyle='--')
plt.tight_layout()
plt.show()

print("Топ-10 признаков с самой сильной связью (по модулю):")
print(corr_with_target[top_20_features])
```

Выводы по диаграмме:

Огромный перевес отрицательной корреляции. 18 из 20 признаков имеют отрицательную связь. Это означает, что при атаках эти значения падают по сравнению с нормальным трафиком. В основном это признаки, связанные с размерами пакетов и сегментов (bwd_seg_size_avg -0.72, bwd_pkt_len_mean -0.72, fwd_pkt_len_max -0.72). В нормальном трафике (BENIGN) серверы передают крупные файлы, а атаки - это мелкие служебные пакеты.

Признаки объёма и окон. Сюда же попали totlen_bwd_pkts (общий объём байт от сервера), down_up_ratio (перекос скачанного/отправленного) и init_bwd_win_byts (размер TCP-окна). Всё это падает при атаках.

Положительная связь. flow_iat_min (+0.54) и byte_ratio (+0.57) - единственные, кто растет при атаке. Это логично: атаки генерируют пакеты с минимальными задержками (высокая частота) и создают сильный перекос в передаче байт (односторонний трафик).
